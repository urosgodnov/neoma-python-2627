## A3 · Lists, dictionaries and indexing from 0

| R | Python | Holds |
|---|---|---|
| `c(42, 38, 51)` | `[42, 38, 51]` (list) | an ordered collection |
| `list(id = 5001, paid = FALSE)` | `{"id": 5001, "paid": False}` (dict) | named values |
| `x[1]` is the **first** element | `x[0]` is the **first** element | |
| `x[-1]` drops the first | `x[-1]` is the **last** | |

````python
monthly_sales = [42, 38, 51, 47, 60, 55]
monthly_sales[0], monthly_sales[-1], monthly_sales[1:3]
````

🔮 **Predict:** how many elements does `monthly_sales[1:3]` return? Slices include the start and **exclude** the end.

````python
len(monthly_sales), sum(monthly_sales), max(monthly_sales)
````

⚠️ Lists are **not** vectors. `monthly_sales * 2` repeats the list. For vector maths we use **NumPy** and **pandas**.

````python
print(monthly_sales * 2)
print(np.array(monthly_sales) * 2)
````

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = pd.Series(monthly_sales, index=months)   # a named vector, like c(Jan = 42, ...)
sales
````

````python
print("Total:", sales.sum(), "| Mean:", round(sales.mean(), 1), "| Best month:", sales.idxmax())
sales[sales > sales.mean()]
````

## B1 · Control flow and comprehensions

Python uses **indentation** (4 spaces) instead of `{ }`. The colon `:` opens a block.

````python
mass = 4200

if mass < 3500:
    label = "light"
elif mass <= 4750:
    label = "medium"
else:
    label = "heavy"

label
````

````python
for month, value in zip(months, monthly_sales):
    flag = "⬆️" if value > 50 else ""
    print(f"{month}: {value} {flag}")
````

A **list comprehension** builds a new list in one line: `[expression for item in collection if condition]`.

````python
gross = [round(p * 1.2, 2) for p in [10, 19.9, 99.99]]
big_months = [m for m, v in zip(months, monthly_sales) if v > 50]
gross, big_months
````

## B2 · Functions

🆚 `vat_price <- function(net, rate = 0.20) {...}` in R becomes `def` in Python. The **docstring** explains the function; **type hints** document the inputs.

````python
def vat_price(net: float, rate: float = 0.20) -> float:
    """Return the gross price, rounded to 2 decimals."""
    return round(net * (1 + rate), 2)


vat_price(10), vat_price(99.99), vat_price(10, rate=0.055)
````

🔮 **Predict:** what happens with `vat_price("10")`? Run it. We will make functions fail **clearly** in Session 2.

````python
try:
    vat_price("10")
except TypeError as e:
    print("TypeError:", e)
````

Functions work on whole columns when the operations inside are vectorised:

````python
vat_price(pd.Series([10, 99.99, 250]))
````

### Missing values
🆚 R's `NA` is `NaN` (numbers) or `None`/`<NA>` in pandas. **Difference:** pandas skips missing values by default. `mean(c(4, NA, 8))` is `NA` in R, but `pd.Series([4, None, 8]).mean()` is `6.0` in pandas.

````python
penguins.isna().sum()
````

````python
penguins["body_mass"].mean(), penguins["body_mass"].mean(skipna=False)
````

### New columns
`pd.cut()` turns numbers into classes, like `case_when()` for ranges. Missing stays missing.

````python
penguins["mass_class"] = pd.cut(
    penguins["body_mass"],
    bins=[0, 3500, 4750, np.inf],
    labels=["light", "medium", "heavy"],
    right=False,
)
penguins[["species", "body_mass", "mass_class"]].sample(5, random_state=1)
````

````python
pd.crosstab(penguins["species"], penguins["mass_class"])
````

### The penguin pricing desk 🐧💶
A (fictional) zoo pays 4 € per 100 g of body mass, that is 0.04 € per gram. Write the price as a function and summarise by species.

````python
def price_penguin(mass_g, rate=0.04):
    """Price in euros: rate × body mass in grams."""
    return round(mass_g * rate, 2)


penguins["price"] = price_penguin(penguins["body_mass"])
penguins.groupby("species")["price"].mean().sort_values(ascending=False).round(2)
````

🆚 The last line in R:
```r
penguins |> summarise(price = mean(price, na.rm = TRUE), .by = species) |> arrange(desc(price))
```
In Session 3 you will learn pandas **method chaining**, Python's version of `|>`.

````python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
ax = sns.boxplot(data=penguins, x="species", y="price", hue="species")
ax.set(title="Gentoo penguins are the most expensive", xlabel="", ylabel="Price (€)")
plt.show()
````
