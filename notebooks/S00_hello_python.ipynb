## 1 · My setup

````python
# ⚙️ Setup: run this cell first (works in Google Colab and in Positron)
from pathlib import Path
import numpy as np
import pandas as pd

REPO = "https://raw.githubusercontent.com/urosgodnov/neoma-python-2627/main/"


def data_path(name):
    """Use the local course folder if you have it, otherwise read the file from GitHub."""
    for base in (Path("data"), Path("../data")):
        if (base / name).exists():
            return base / name
    return REPO + "data/" + name


pd.set_option("display.max_columns", 30)
print("Python ready · pandas", pd.__version__)
````

````python
import sys
import platform

import matplotlib
import seaborn as sns
import statsmodels

print("Python     ", sys.version.split()[0])
print("Platform   ", platform.system())
print("pandas     ", pd.__version__)
print("matplotlib ", matplotlib.__version__)
print("seaborn    ", sns.__version__)
print("statsmodels", statsmodels.__version__)
````

✅ **Check:** you see Python 3.11 or newer and five library versions. An `ImportError` means a package is missing: in Positron run `pip install pandas seaborn statsmodels` in the Terminal (see the install guide in Moodle). In Colab everything is pre-installed.

## 2 · The data you will meet all semester

The **same datasets as in the R course**: Palmer penguins, monthly sales files, NYC flights, Gapminder, Ames house prices and an HR dataset. Same data, new language: you will see how the same question is answered in R and in Python.

````python
penguins = pd.read_csv(data_path("penguins.csv"))
penguins.head()
````

````python
print(f"The data contains {len(penguins)} penguins from {penguins['species'].nunique()} species.")
````

## 3 · A first chart

````python
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
ax = sns.scatterplot(data=penguins, x="flipper_len", y="body_mass", hue="species", alpha=0.7)
ax.set(
    title="Heavier penguins have longer flippers",
    xlabel="Flipper length (mm)",
    ylabel="Body mass (g)",
)
plt.show()
````
